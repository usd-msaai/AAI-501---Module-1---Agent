# Connect the Gemini API


```python
from google.colab import userdata
```

# Connect API

```python
# Load the API key from Colab secrets
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
```

# Build a basic langchain agent

```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
from langchain_google_genai import ChatGoogleGenerativeAI
from langgraph.prebuilt import create_react_agent

#Instantiate the Wikipedia tool
api_wrapper = WikipediaAPIWrapper(top_k_results=1, doc_content_chars_max=1000)
wikipedia_tool = WikipediaQueryRun(api_wrapper=api_wrapper)

#Initialize the Gemini model, explicitly passing the API key and using a supported model
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash", temperature=0, google_api_key=GOOGLE_API_KEY)

#Initialize the LangChain agent
from langchain.agents import create_agent  # langchain / LangGraph 1.x
agent = create_agent(llm, tools=[wikipedia_tool])
```

# Run the agent

```python
# Run the agent with a sample query
result = agent.invoke({"messages": [("user", "What is the capital of France?")]})

# Print the full trace (what verbose=True used to show), then the final answer
for m in result["messages"]:
    m.pretty_print()
result["messages"][-1].content
```
